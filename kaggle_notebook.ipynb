## 7. Display results

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

results_dir = "/kaggle/working/results"
pngs = [f for f in os.listdir(results_dir) if f.endswith(".png")] if os.path.isdir(results_dir) else []

if pngs:
    fig, axes = plt.subplots(1, len(pngs), figsize=(6 * len(pngs), 6))
    if len(pngs) == 1:
        axes = [axes]
    for ax, png in zip(axes, pngs):
        ax.imshow(Image.open(os.path.join(results_dir, png)))
        ax.set_title(png)
        ax.axis("off")
    plt.tight_layout()
    plt.show()
else:
    print("No result images found — run inference first.")
```
